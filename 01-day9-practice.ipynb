## 1 — Compound Return Function

**Compound interest** means earning interest on both the principal and previously earned interest.

Formula: **A = P × (1 + r)ⁿ**

In finance, this is how investments grow over time — the longer the horizon, the more powerful the compounding effect.

```python
def compound_return(prin, ann_rate, years):
    amount = prin * (1 + ann_rate) ** years
    print(f'Rs {prin:,} at {ann_rate*100}% for {years} years → Rs {amount:,.2f}')
    return round(amount, 2)
```

```python
compound_return(10000, 0.1, 5)
```

## 2 - Daily Returns 

**Daily Returns** talks about how much a stock price has gone up or down from previous day. 

**Std. Deviation** talks about fluctuations in daily prices of stocks. It gives us idea about how volatile a stock is

When std deviation is much larger than the mean return, day-to-day swings can easily wipe out several days of gains — indicating a risky stock.

```python
import numpy as np

prices     = np.array([100, 102, 99, 105, 103])

daily_ret = (prices[1:] - prices[:-1]) / prices[:-1]
mean_ret = np.mean(daily_ret)
std_dev = np.std(daily_ret)

print(f'Daily Returns: {np.round(daily_ret, 4)}')
print(f'Mean Return  : {mean_ret:.4f}')
print(f'Std Deviation: {std_dev:.4f}')
```

## 3 - PE Ratio

Price to Earning **(PE)** ratio talks about how much investors are willing to pay for each rupee earned by company.

A high PE means investors expect strong future growth but are paying a premium. A low PE may indicate an undervalued stock or a business with weak growth prospects

```python
import pandas as pd

data = {
    'Ticker': ['RELIANCE', 'TCS', 'INFY', 'HDFCBANK', 'ITC'],
    'Price':  [2450, 3800, 1500, 1600, 420],
    'Sector': ['Energy', 'IT', 'IT', 'Banking', 'FMCG'],
    'PE_Ratio': [22.5, 28.3, 24.1, 18.7, 26.4]
}
df = pd.DataFrame(data)
print(df)
print(df.loc[df['Sector'] == 'IT'])
print(df.loc[df['PE_Ratio'].idxmax(), 'Ticker'])
df['Valuation'] = df['PE_Ratio'].map(lambda x: 'Expensive' if x > 25 else 'Fair Value')
print(df)
```

## 4 - PCT Change and Resample

**pct_change()** calculates the percentage change in stock price relative to the previous day

**resample('ME')** groups time series data into monthly buckets, allowing us to compute monthly aggregates like average return

As an analyst, monthly average return tells us how a stock performed on average each month, helping identify seasonality or event-driven trends

```python
import yfinance as yf

returns = yf.download('TCS.NS', start = '2025-05-20', end = '2026-05-20')['Close']
print (returns)

daily_ret = returns.pct_change().dropna()
print(daily_ret)

ret = daily_ret['TCS.NS']  # extract as Series
print(f'Best day:  {ret.idxmax().date()} → {ret.max():.4f}')
print(f'Worst day: {ret.idxmin().date()} → {ret.min():.4f}')

monthly_avg = ret.resample('ME').mean()
print(monthly_avg.round(4))
```

## 5 - Why 252
While calculating **Annualised Return** we multiply with 252 because on an average there are 252 trading days in a year.

While calculating **Annualised Volatility** we multiply with sqr root of 252 because Volatility (std deviation) scales with the square root of time — so to annualise daily volatility we multiply by √252, not 252 itself.

```python
ann_ret = ret.mean() * 252
ann_vol = ret.std() * (252 ** 0.5)
rf_daily = 0.065 / 252
sharpe_rat = (ret.mean() - rf_daily) / ret.std()

print(f'Annualised Return for TCS is : {ann_ret*100:.2f}%')
print(f'Annualised Volatility for TCS is : {ann_vol*100:.2f}%')
print(f'Sharpe Ratio for TCS is : {sharpe_rat:.2f}')
```
